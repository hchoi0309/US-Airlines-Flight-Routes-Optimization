# Sample Airline Information Data

To run our optimization model and evaluate its performance, we need to compare the results against actual data. In this project, we will use information from Allegiant Airlines, which includes their flight routes, fleet information, and historical profit data.

```python
import pandas as pd
```

```python
allegiant_df = pd.read_csv("../data/allegiant_routes_info.csv")
display(allegiant_df.head())
```

```python
# Load historical flight data that has been preprocessed
historical_df = pd.read_csv("../data/cleaned_data.csv")
# First aggregate historical_df to get unique airport pairs
unique_routes_df = historical_df.groupby(['airport_1', 'airport_2']).agg({
    'state_1': 'first',
    'city_1': 'first',
    'state_2': 'first',
    'city_2': 'first',
    'population_1': 'mean',
    'density_1': 'mean',
    'population_2': 'mean',
    'density_2': 'mean',
    'lat_1': 'first',
    'lon_1': 'first',
    'lat_2': 'first',
    'lon_2': 'first'
}).reset_index()
print("There are a total of", unique_routes_df.shape[0], "unique routes.")
```

```python
# Merge the dataframes on airport_1 and airport_2
merged_df = pd.merge(
    allegiant_df,
    unique_routes_df,
    on=['airport_1', 'airport_2'],
    how='left'  # Use 'left' to keep all rows from df
)

# Display the merged dataframe
display(merged_df.head())
```
